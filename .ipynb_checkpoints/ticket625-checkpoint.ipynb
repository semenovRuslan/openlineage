```python
import json

class DataElement:
    def __init__(self, field_name: str, field_type: str):
        self.field_name = field_name
        self.field_type = field_type

    def to_dict(self) -> dict:
        return {
            "field_name": self.field_name,
            "field_type": self.field_type
        }
class OpenLineage:
    def __init__(self):
        self._columns = []

    def add_column(self, element: DataElement) -> None:
        if not isinstance(element, DataElement):
            raise TypeError(f"Expected DataElement object, got {type(element).__name__}")
        
        self._columns.append(element)

    def print_as_json(self) -> None:
        serialized_list = [col.to_dict() for col in self._columns]
        ol_document = json.dumps(serialized_list, ensure_ascii=False, indent=2)
        return ol_document

    def __len__(self) -> int:
        return len(self._columns)
```

```python
# 1. Инициализация объекта
manager = OpenLineage()

raw_metadata = [
    ("user_id", "INT"),
    ("email", "VARCHAR"),
    ("updated_at", "TIMESTAMP")
]

# 2. Цикл внешнего приложения
for name, f_type in raw_metadata:
    # Создаем экземпляр нашего класса без dataclasses
    element = DataElement(field_name=name, field_type=f_type)
    
    # Добавляем в менеджер
    manager.add_column(element)

# 3. После окончания цикла выводим результат в JSON
print(manager.print_as_json())
```

```python
ol_document = json.dumps(["user_id", "user_name", "email", "created_at"], ensure_ascii=False)
print (ol_document)
```
