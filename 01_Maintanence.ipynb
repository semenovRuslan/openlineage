initianally before we started Jupiter Notebook we should configure our terminal:
1. conda config --add channels conda-forge
2. conda create --name openlineage python==3.10 python-dotenv
3. conda init
4. source  ~/.bashrc
5. conda activate openlineage
альтернатива (3, 4, 5) сразу >source activate openlineage

теперь установим в нашем environment jupiter server
1. conda install ipykernel jupyter jupyterlab
2. python -m ipykernel install --user --name=openlineage --display-name "openlineage testing environment"

теперь установим библиотеки openlineage
1. pip install openlineage-python==1.16.0
2. pip install numpy
3. pip install openlineage-sql

теперь запускаем jupiter и работаем там с нашим openlinege потоком (kernel)
1. jupyter lab

выключение среды 
1. conda deactivate
