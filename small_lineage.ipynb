```python
import openlineage_sql

# Базовый пример парсинга SQL-запроса
sql_query = "INSERT INTO target_db.users_summary SELECT id, name FROM source_db.users"

# Функция parse принимает список запросов
metadata = openlineage_sql.parse([sql_query])

# Извлекаем lineage данных
print("Входные таблицы:", metadata.in_tables)
print("Выходные таблицы:", metadata.out_tables)
```
