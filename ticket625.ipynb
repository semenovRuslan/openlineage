```python
import json

class DataElement:
    def __init__(self, field_name: str, field_type: str):
        self.field_name = field_name
        self.field_type = field_type

    def to_dict(self) -> dict:
        return {
            "field_name": self.field_name,
            "field_type": self.field_type
        }
class OpenLineage:
    def __init__(self):
        self._tables = {}

    def add_column(self, table_name: str, element: DataElement) -> None:
        if not isinstance(element, DataElement):
            raise TypeError(f"Expected DataElement object, got {type(element).__name__}")
        
        if table_name not in self._tables:
            self._tables[table_name] = []
            
        # Добавляем колонку в соответствующую таблицу
        self._tables[table_name].append(element)

    def print_as_json(self) -> None:
        serialized_output = {}
        
        # Проходим по всем таблицам и сериализуем их колонки
        for table_name, columns in self._tables.items():
            serialized_output[table_name] = [col.to_dict() for col in columns]
        
        # Формируем итоговый JSON-документ
        ol_document = json.dumps(serialized_output, ensure_ascii=False, indent=2)
        return ol_document

    def __len__(self) -> int:
        return len(self._columns)
```

```python
# 1. Инициализация объекта
manager = OpenLineage()

# 2. Внешний цикл имитирует хаотичное поступление колонок из разных таблиц
manager.add_column("users", DataElement("user_id", "INT"))
manager.add_column("orders", DataElement("order_id", "BIGINT"))
manager.add_column("users", DataElement("email", "VARCHAR"))
manager.add_column("orders", DataElement("amount", "DECIMAL"))
manager.add_column("users", DataElement("created_at", "TIMESTAMP"))

# 3. Печать структурированного отчета
# 3. После окончания цикла выводим результат в JSON
print(manager.print_as_json())
```
